## The Core Rule

Two boolean expressions are equivalent only when they produce the **same result for every possible input**, confirmed with a full truth table, never a single test case.

```python
public class QuickCheck {
    public static void main(String[] args) {
        boolean a = true, b = false;
        System.out.println(!(a && b));   // true
        System.out.println(!a || !b);    // true — matches
    }
}

QuickCheck.main(null);
```

## Boolean Operators Reference

| Operator | Meaning | CSA usage |
|---|---|---|
| `==` | Equal to | Compare primitive values |
| `!=` | Not equal to | Compare primitive values |
| `&&` | Logical AND | True only if both sides are true |
| `\|\|` | Logical OR | True if at least one side is true |
| `!` | Logical NOT | Flips true to false and false to true |

## Equivalence & De Morgan's Grammar

| Rule | Use |
|---|---|
| Equivalent expressions | Two expressions that match on **every row** of the truth table |
| Truth table | A chart testing every possible T/F input combination |
| De Morgan's Law 1 | `!(a && b)` is equivalent to `!a \|\| !b` |
| De Morgan's Law 2 | `!(a \|\| b)` is equivalent to `!a && !b` |
| Short-circuit evaluation | Java skips checking the right side once the result is already known |

## 1. LxD Cycle Process

**Empathize:** I noticed a lot of students check whether two boolean expressions are equivalent by testing just one case (like `a = true, b = true`) and assuming that if it matches, the expressions are always equal. This creates a shaky mental model and leads to real mistakes on the AP exam's equivalence questions, where three of four answer choices "look" equivalent but fail on exactly one row.

**Define:**

- **POV:** AP CSA students need a reliable method to compare boolean expressions because testing a single case creates false confidence and leads to wrong answers.
- **Learning Goal:** Students will determine whether two boolean expressions are equivalent by building and comparing complete truth tables, and will apply De Morgan's Laws to rewrite negated compound expressions.

**Ideate:**

- **HMW Question:** How might we get students to build a full truth table instead of testing just one case?
- **Activity:** Given two boolean expressions side by side, students predict the truth table by hand, then verify it with a Java program that loops through every T/F combination.

**Prototype & Test:** I ran this activity with my project team. They found working out a full 3-variable, 8-row truth table by hand too slow for a 2-minute in-class task, so I moved the 3-variable version to homework and kept the in-class hack to a simpler 2-variable check (documented below).

## 2. Lesson Plan

**Learning Objective:** By the end of this lesson, you will be able to determine whether two boolean expressions are equivalent using a truth table, and rewrite negated compound expressions using De Morgan's Laws.

**Success Criteria:** You can build a complete truth table for a boolean expression, compare it row-by-row against another expression, and correctly identify whether they are equivalent, before ever running any code.

### Tech Talk (3 minutes)

We are comparing boolean expressions. This means you can't plug in one value and call it done. You need to test **every possible combination** of inputs and confirm the results always match.

- **The Rule:** Two expressions are equivalent only if they agree on every single row of the truth table.
- ✅ **Do this:** build all 4 rows for a 2-variable expression before concluding equivalence
- ❌ **Don't do this:** test only `a=true, b=true` and assume the rest follow

Why do we do this? The AP exam tests equivalence with multiple-choice questions designed so a partial check picks the wrong answer. A full truth table is the only reliable proof.

### Code Examples

#### A. Simple: Testing One Boolean Expression

```python
public class BooleanBasics {
    public static void main(String[] args) {
        boolean a = true;
        boolean b = false;
        System.out.println(a && b);   // false
        System.out.println(a || b);   // true
        System.out.println(!a);       // false
    }
}

BooleanBasics.main(null);
```

#### B. Intermediate: Comparing Two Expressions Across All Combinations

```python
public class CompareTwo {
    public static void main(String[] args) {
        boolean[] values = {true, false};
        for (boolean a : values) {
            for (boolean b : values) {
                boolean left = !(a && b);
                boolean right = !a || !b;
                System.out.println("a=" + a + " b=" + b + " -> " + left + " | " + right);
            }
        }
    }
}

CompareTwo.main(null);
```

#### C. Complex: Three-Variable Equivalence Check

```python
public class ThreeVariableCheck {
    public static void main(String[] args) {
        boolean[] values = {true, false};
        for (boolean a : values) {
            for (boolean b : values) {
                for (boolean c : values) {
                    boolean left = a && (b || c);
                    boolean right = (a && b) || (a && c);
                    System.out.println("a=" + a + " b=" + b + " c=" + c +
                        " -> " + left + " | " + right);
                }
            }
        }
    }
}

ThreeVariableCheck.main(null);
```
