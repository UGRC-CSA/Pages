## The Core Rule

Combine simple boolean expressions with `&&`, `\|\|`, and `!` to test multiple conditions at once, always evaluated in a fixed order: `!` first, then `&&`, then `\|\|`.

````java
public class QuickCheck {
    public static void main(String[] args) {
        int age = 17;
        boolean hasPermit = true;

        boolean canDriveWithSupervision = (age >= 15) && hasPermit;
        System.out.println(canDriveWithSupervision);   // true
    }
}

QuickCheck.main(null);
````

## Compound Operators Reference

| Operator | Meaning | CSA usage |
|---|---|---|
| `&&` | AND | True only when **both** sides are true |
| `\|\|` | OR | True when **at least one** side is true |
| `!` | NOT | Flips true to false and false to true |
| `( )` | Grouping | Forces a specific evaluation order |

## Precedence & Range-Check Grammar

| Rule | Use |
|---|---|
| Precedence order | `!` evaluates before `&&`, which evaluates before `\|\|` |
| Range check | Combine two comparisons with `&&` to test "between" (e.g. `x >= 0 && x <= 10`) |
| Either-or check | Combine two comparisons with `\|\|` to test "any of these" |
| Parentheses | Use them to force your intended grouping instead of relying on default precedence |

## 1. LxD Cycle Process

**Empathize:** I noticed a lot of students write compound conditions like `age >= 13 && <= 19` (missing the second variable) or `score > 90 || score > 80 && passed` without parentheses, and get confused when the result doesn't match what they expected. They don't yet have a mental model for how Java actually groups `&&` and `\|\|` together.

**Define:**

- **POV:** AP CSA students need a reliable way to combine multiple conditions because writing them without understanding precedence produces bugs that are hard to trace.
- **Learning Goal:** Students will correctly combine boolean expressions using `&&`, `\|\|`, and `!`, and will use parentheses to control evaluation order instead of relying on default precedence.

**Ideate:**

- **HMW Question:** How might we get students to always double-check precedence instead of assuming left-to-right reading order?
- **Activity:** Given a compound expression with no parentheses, students first predict how Java groups it, then add parentheses to make their intended grouping explicit and check the output matches.

**Prototype & Test:** I ran this with my project team. They mixed up range checks and either-or checks when I combined both patterns in one example, so I split range checks and either-or checks into two separate, clearly labeled code examples (documented below).

## 2. Lesson Plan

**Learning Objective:** By the end of this lesson, you will be able to combine boolean expressions using `&&`, `\|\|`, and `!`, and correctly predict how Java evaluates a compound expression based on operator precedence.

**Success Criteria:** You can write a compound boolean expression for a given rule (a range, an either-or condition, or a negated condition) and correctly predict its result without running the code first.

### Tech Talk (3 minutes)

A single comparison can only ask one question. Compound expressions let you ask several questions at once and combine the answers.

- **The Rule:** `!` runs first, then `&&`, then `\|\|`. When in doubt, add parentheses.
- ✅ **Do this:** `(age >= 13) && (age <= 19)`
- ❌ **Don't do this:** `age >= 13 && <= 19` (this isn't even valid Java, but it's a common typo pattern)

Why do we do this? The AP exam frequently gives you a compound expression with no parentheses and asks you to trace through it using precedence rules alone. Getting the order wrong is one of the most common exam mistakes.

### Code Examples

#### A. Simple: Combining Two Conditions with AND

````java
public class RangeCheck {
    public static void main(String[] args) {
        int score = 85;
        boolean isBGrade = (score >= 80) && (score < 90);
        System.out.println(isBGrade);   // true
    }
}

RangeCheck.main(null);
````

#### B. Intermediate: Combining Conditions with OR

````java
public class EitherOrCheck {
    public static void main(String[] args) {
        String day = "Saturday";
        boolean isWeekend = day.equals("Saturday") || day.equals("Sunday");
        System.out.println(isWeekend);   // true

        int temperature = 102;
        boolean needsAttention = temperature < 95 || temperature > 100;
        System.out.println(needsAttention);   // true
    }
}

EitherOrCheck.main(null);
````

#### C. Complex: Precedence Without Parentheses

````java
public class PrecedenceDemo {
    public static void main(String[] args) {
        boolean a = true, b = false, c = false;

        // No parentheses: && binds tighter than ||
        // Java reads this as: a || (b && c)
        boolean noParens = a || b && c;
        System.out.println("a || b && c -> " + noParens);

        // Explicit grouping showing the same evaluation order
        boolean explicit = a || (b && c);
        System.out.println("a || (b && c) -> " + explicit);

        // A different grouping changes the result
        boolean differentGrouping = (a || b) && c;
        System.out.println("(a || b) && c -> " + differentGrouping);
    }
}

PrecedenceDemo.main(null);
````
