## The Core Rule

An `else if` chain evaluates conditions **top to bottom** and runs only the code under the *first* one that is true — every later branch is skipped, even one that would also be true on its own. Order is part of the logic, not just style.

**Learning objective:** Trace and write `if / else if / else` decision chains to determine which single branch executes for a given input.

**Success criteria — I can:**
- Trace which branch of an `else if` chain runs for a given input.
- Explain why only one branch in the chain executes, even when more than one condition would be true on its own.
- Write my own `else if` chain that classifies a value into categories.

*(Engage: predict the output below before you run it — this takes under five minutes.)*

```java
// CODE_RUNNER: Predict the output for myAge = 17, then change myAge and rerun.
public class Main {
    public static void main(String[] args) {
        int myAge = 17; // adjust the age to see the difference in what it prints out
        System.out.println("Current age: " + myAge);
        if (myAge >= 18)
        {
            System.out.println("You can register to vote.");
            System.out.println("You are old enough for a license to drive.");
        }
        else if (myAge >= 16) {
            System.out.println("You are old enough for a license to drive.");
        }
        else if (myAge >= 15) {
            System.out.println("You can learn to drive a car.");
        }
        else if (myAge >= 14) {
            System.out.println("You are 14");
        }
    }
}

Main.main(null);
```

### Reference Guide

#### Chain Evaluation Rules

| Rule | What it means |
|---|---|
| Order matters | The first true condition runs; everything after it is skipped, even a later condition that is also true. |
| Put the narrowest condition first | A broad condition (like `age >= 0`) placed early will "shadow" — permanently hide — a narrower condition placed after it. |
| `else` is an optional catch-all | It runs only when every condition above it was false; it does not need its own condition. |
| Only one branch's body runs | Even though multiple `if`/`else if` conditions might independently be true, the chain stops at the first match. |

#### Common Nested-If Mistakes

| Mistake | Why it breaks |
|---|---|
| Writing the broadest range first | e.g. checking `score >= 60` before `score >= 90` means no student can ever reach the `90` branch. |
| Forgetting `else` and repeating the same variable in separate, un-chained `if` statements | Removes the "only one branch runs" guarantee — now more than one block can execute. |
| Assuming Java checks every condition | Java stops at the first `true` — later conditions are never even evaluated. |

### Tech Talk (3 minutes)

**Else if statements** let you check multiple conditions in sequence when only one outcome should apply.

- **The Rule:** each condition is evaluated in the order written; the first true condition's code runs, and the rest — even ones that are also true — are skipped.
- ✅ **Do this:** put the narrowest, most specific condition first (`age >= 18` before `age >= 16`).
- ❌ **Don't do this:** put a broad condition first (`age >= 0` before `age >= 18`) — it will shadow every narrower condition below it.

Why do we do this? On the AP exam and in real code, a chain that "looks right" can silently return the wrong branch if the conditions are in the wrong order — there's no error message, just a wrong answer.

### Raise your hand:

### Where else do you see this?

A backend API has to pick exactly one response for a request — it can't return `200 OK` *and* `404 Not Found` for the same call. That's the same "first true condition wins" shape as an `else if` chain: check the most specific problem first (does the item exist?), then broader ones, and let one final `else` catch everything else.

```java
if (item == null) {
    return notFound();          // most specific: does it exist at all?
}
else if (quantity > maxAllowed) {
    return badRequest();        // narrower rule checked before the general case
}
else {
    return ok(item);            // catch-all: nothing above matched
}
```

### Code Examples

#### A. Simple: Classifying One Value into Non-Overlapping Bands

```java
public class TicketPrice {
    public static void main(String[] args) {
        int age = 8;

        if (age < 3) {
            System.out.println("Ticket price: free");
        }
        else if (age <= 12) {
            System.out.println("Ticket price: $8 (child)");
        }
        else if (age < 65) {
            System.out.println("Ticket price: $14 (adult)");
        }
        else {
            System.out.println("Ticket price: $10 (senior)");
        }
    }
}

TicketPrice.main(null);
```

#### B. Intermediate: Why Condition Order Can Shadow a Branch

Both chains below test the exact same weight against the exact same three rules — only the *order* is different.

```java
public class ShippingTier {
    public static void main(String[] args) {
        double weight = 2.5; // pounds

        // WRONG ORDER: the broad ">= 0" check shadows every narrower tier below it
        String wrongTier = "";
        if (weight >= 0) { wrongTier = "standard"; }
        else if (weight <= 1) { wrongTier = "envelope"; }
        else if (weight <= 5) { wrongTier = "small box"; }

        // CORRECT ORDER: narrowest, most specific condition goes first
        String rightTier = "";
        if (weight <= 1) { rightTier = "envelope"; }
        else if (weight <= 5) { rightTier = "small box"; }
        else if (weight >= 0) { rightTier = "standard"; }

        System.out.println("Wrong order  -> " + wrongTier);  // standard, even though 2.5 lb fits "small box"
        System.out.println("Right order  -> " + rightTier);  // small box
    }
}

ShippingTier.main(null);
```

Notice what makes this dangerous: `weight >= 0` is true for *every* package, so once it's checked first, none of the more specific branches below it can ever run. The chain still compiles, still prints an answer, and gives no error — it's just the wrong one.

#### C. Complex: `else if` Chain vs. `switch`

Both versions below decide a price from the exact same `size` variable. A `switch` is a good fit **only** when every branch tests one variable against exact, discrete values — it cannot express a range or inequality like `age >= 18`, which is why the age-eligibility and shipping-tier examples above have to stay as `else if` chains.

```java
public class SizePrice {
    public static void main(String[] args) {
        String size = "L";

        // Version 1: else-if chain -- works, but every branch re-checks the same variable for equality
        double priceIfChain;
        if (size.equals("S")) { priceIfChain = 3.50; }
        else if (size.equals("M")) { priceIfChain = 4.00; }
        else if (size.equals("L")) { priceIfChain = 4.50; }
        else { priceIfChain = 5.00; } // "XL"
        System.out.println("if/else-if price: " + priceIfChain);

        // Version 2: switch -- fits better here because every branch tests the SAME variable
        // against exact, discrete values (not a range/inequality like age or grade)
        double priceSwitch;
        switch (size) {
            case "S":
                priceSwitch = 3.50;
                break;
            case "M":
                priceSwitch = 4.00;
                break;
            case "L":
                priceSwitch = 4.50;
                break;
            default:
                priceSwitch = 5.00; // "XL"
        }
        System.out.println("switch price:     " + priceSwitch);
    }
}

SizePrice.main(null);
```
